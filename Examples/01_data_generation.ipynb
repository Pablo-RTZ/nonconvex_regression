```python
import matplotlib.pyplot as plt
import numpy as np

from nonconvex_regression import clean_data, outlier_data

plt.style.use("seaborn-v0_8-whitegrid")
```

```python
n = int(1e3)
beta = np.array([1.0, 1.0])
epsilon = 1e-1
x_min, x_max = -2, 2
p = 0.1
A = 10

points_clean = clean_data(n, beta, epsilon, x_min, x_max)
points_outlier = outlier_data(n, beta, epsilon, x_min, x_max, p, A)

x_line = np.linspace(x_min, x_max, 200)
true_line = beta[0] + beta[1] * x_line
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5), constrained_layout=True)
fig.suptitle("Linear data generation with outliers", fontsize=16, y=1.02)

for ax in axes:
    ax.plot(x_line, true_line, color="black", lw=2, ls="--", label="True model")
    ax.set_xlim(x_min, x_max)
    ax.set_xlabel("x", fontsize=11)
    ax.set_ylabel("y", fontsize=11)
    ax.grid(alpha=0.35)

axes[0].scatter(
    points_clean[:, 0],
    points_clean[:, 1],
    s=18,
    alpha=0.75,
    c="#4C72B0",
    edgecolors="white",
    linewidths=0.35,
    label="Clean observations",
)
axes[0].set_title("Clean data", fontsize=12)

axes[1].scatter(
    points_outlier[:, 0],
    points_outlier[:, 1],
    s=18,
    alpha=0.75,
    c="#C44E52",
    edgecolors="white",
    linewidths=0.35,
    label="Contaminated data",
)
axes[1].set_title("Data with outliers", fontsize=12)

for ax in axes:
    ax.legend(frameon=True, fancybox=True, framealpha=0.9)

plt.show()
```
